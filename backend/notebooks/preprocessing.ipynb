```python
#drop course_description which has null values since it is important for understanding about the course
df = df.dropna(subset=["course_description"])
```

```python
df.isnull().sum()
```

```python
df["course_students_enrolled"] = df["course_students_enrolled"].fillna("0")
```

```python
df["course_students_enrolled"] = (
    df["course_students_enrolled"]
    .astype(str)
    .str.replace(",", "")
    .astype(float)
)
```

```python
df.info()
```

```python
#fix the review count column
def convert_k(x):
    if isinstance(x, str):
        x = x.replace("K", "")
        return float(x) * 1000
    return x

df["Review Count"] = df["Review Count"].apply(convert_k)
```

```python
df.head(5)
```

```python
#clean skills field and make it in the form of list
df["Skills"] = df["Skills"].apply(lambda x: [s.strip() for s in x.split(",")])
```

```python
df.head()
```

```python
#normalize text fields
df["Title"] = df["Title"].str.lower()
df["Organization"] = df["Organization"].str.lower()
df["course_description"] = df["course_description"].str.lower()
```

```python
df.head(5)
```

```python
#remove the unnamed column
df = df.drop(columns=["Unnamed: 0"])
```

```python
#find the unique values in difficulty column
df["Difficulty"].unique()
```

```python
#standardize the difficulty column
df["Difficulty"] = (
    df["Difficulty"]
    .str.strip()
    .str.lower()
)
```

```python
mapping = {
    "beginner": "beginner",
    "intermediate": "intermediate",
    "advanced": "advanced",
    "mixed": "all_levels"
}

df["Difficulty"] = df["Difficulty"].replace(mapping)
```

```python
#type column unique values
df["Type"].unique()
```

```python
#standardize the type column
df["Type"] = df["Type"].str.strip().str.lower()
```

```python
type_mapping = {
    "professional certificate": "professional certificate",
    "specialization": "specialization",
    "course": "course"
}

df["Type"] = df["Type"].replace(type_mapping)
```

```python
df.to_csv("../data/processed/cleaned_courses.csv", index=False)
```
